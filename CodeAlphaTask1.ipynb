```python
df.groupby('Admission Type')['Billing Amount'].mean()
```

```python
#statistical Test
from scipy.stats import ttest_ind
male_bill = df[df['Gender']== 'Male']['Billing Amount']
female_bill = df[df['Gender']=='Female']['Billing Amount']
t_stat,p_val = ttest_ind(male_bill,female_bill,nan_policy = 'omit')
p_val,t_stat
```

**Insights**
The p-val is 0.285 which is greater than 0.05.
this statistical test proved that there is no significant difference between gender and billing amount.


```python
#Admission type really effects billing amount
from scipy.stats import f_oneway
elective_bill = df[df['Admission Type']=='Elective']['Billing Amount']
emergency_bill = df[df['Admission Type']=='Emergency']['Billing Amount']
urgent_bill = df[df['Admission Type']=='Urgent']['Billing Amount']
f_stat,p_val_anova = f_oneway(elective_bill,emergency_bill,urgent_bill)
f_stat,p_val_anova
```

**INSIGHTS**
The p-val_anova is 0.726 whic is also greater than 0.05
from this test the Admission Type does not effect the Billing Amount.
